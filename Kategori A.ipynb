```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data (perhatikan sep=';')
df = pd.read_csv("Kelas D_Happiness and Satisfaction.csv", sep=';')

# Menghitung jumlah negara tiap region
region_count = df['Region'].value_counts()

# Membuat Pie Chart
plt.figure(figsize=(10,8))
plt.pie(region_count,
        labels=region_count.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title('Distribusi Jumlah Negara Berdasarkan Region')
plt.axis('equal')
plt.show()
```
